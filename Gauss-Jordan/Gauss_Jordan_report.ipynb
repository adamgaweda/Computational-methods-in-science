Przedstawienie algorytmu eliminacji Gaussa-Jordana.
Algorytm w uproszczeniu polega na odejmowaniu wiersza i-tego (1-n) od wierszy j (i-n).
W algebrze zmiennoprzecinkowej może wywołać to czasem błędy reprezentacyjne w wyniku czego nasze wyniki będą zaszumione. W tym celu przeprowadza się pivoting pełny lub częściowy. Pivoting polega na szukaniu elementu o największej wartości (bezwględnej) by operacje odejmowania miały możliwie jak najmniejszy błąd reprezentacyjny. W przypadku partial pivotingu mówimy o znalezieniu wiersza/kolumny, zaczynającej się największej wartości, pełny pivoting szuka w całej macierzy i sprytnie zamienia miejscami wierszy z kolumnami, by macierz zachowała swoje własności

```python
def gauss_jordan_2arg(A, b):
    return gauss_jordan_1arg(np.hstack((A, b.reshape(-1,1))))

def gauss_jordan_1arg(B):
    n = len(B)
    B= B.copy().astype(float)

    for i in range(n):
        max_row = i
        for j in range(i, n):
            if abs(B[j][i]) >= abs(B[max_row][i]):
                max_row = j
        B[[i, max_row]] = B[[max_row, i]]

        for j in range(n):
            if (i == j):
                continue
            if B[i][i]==0:
                print("wywaliło się, słabe dane")
                return B
            a = B[j][i] / B[i][i]
            for k in range(i, n + 1):
                B[j][k] -= (B[i][k] * a)
            B[j][i] = 0

        a = B[i][i]
        for j in range(i, n + 1):
            B[i][j] /= a

    return B[:,-1]
```

Przetestujmy różnice w czasie z funkcjami bibliotecznymi

```python
def main():
    n = 600
    A = np.random.rand(n, n)
    b = np.random.rand(n)

    # Gauss-Jordan
    start = time.time()
    x1 = gauss_jordan_2arg(A.copy(), b.copy())
    t1 = time.time() - start

    # numpy
    start = time.time()
    x2 = np.linalg.solve(A, b)
    t2 = time.time() - start

    print("Gauss-Jordan:", t1)
    print("numpy.linalg.solve:", t2)
    print("błąd:", np.linalg.norm(x1 - x2))

if __name__ == "__main__":
    main()
```

Możemy stwierdzić, że błąd jest marginalny, lecz czas funkcji bibliotecznej powala w porównaniu do funkcji napisanej ręcznie.

Faktoryzacja LU jest jedną z podstawowych faktoryzacji do rozwiązywania układów równań i jedną z najprostszych w zrozumieniu. Zadaniem było napisanie algorytmu, który zrobi faktoryzację LU w miejscu - in-situ.

```python
import numpy as np
import time

def lu_factorization(A):
    n=len(A)

    for j in range(n):
        for i in range(j + 1, n):
            mult = A[i][j] / A[j][j]
            for k in range(j, n):
                A[i][k] -= A[j][k] * mult
            A[i][j] = mult
    return A
```
